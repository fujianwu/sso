```python
import sqlite3

def add_phone_column():
    conn = sqlite3.connect('sso_auth.db')
    cursor = conn.cursor()
    
    try:
        # 1. 先添加普通列（不带 UNIQUE）
        print("正在添加 phone 列...")
        cursor.execute("ALTER TABLE users ADD COLUMN phone TEXT;")
        
        # 2. 为该列创建唯一索引，实现 UNIQUE 约束效果
        print("正在创建唯一索引...")
        cursor.execute("CREATE UNIQUE INDEX idx_users_phone ON users (phone);")
        
        conn.commit()
        print("成功添加 phone 列并设置唯一约束！")
    except sqlite3.OperationalError as e:
        if "duplicate column" in str(e).lower() or "already exists" in str(e).lower():
            print("操作失败：该列或索引可能已经存在。")
        else:
            print(f"发生错误: {e}")
    finally:
        conn.close()

if __name__ == "__main__":
    add_phone_column()
```
